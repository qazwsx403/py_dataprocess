二、欧几里得算法

```python
# 1. gcd 函数
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 2. 评估三组数字
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300)  =", gcd(1200, 300))
print("gcd(5040, 60)   =", gcd(5040, 60))

# 3. 负数情况
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(2018, -2019) =", gcd(2018, -2019))
print("gcd(-12, -8)     =", gcd(-12, -8))
```

当参数为负数时，Python 的 % 运算符返回的结果符号跟除数一致，因此循环仍会终止，返回值的绝对值是两数绝对值的 GCD，但符号可能为负。例如 gcd(-2018, 2019) 返回 -1，gcd(-12, -8) 返回 -4。

三、近似欧拉数 e

```python
import math

# 1. euler_limit：e ≈ (1 + 1/n)^n
def euler_limit(n):
    return (1 + 1 / n) ** n

print("euler_limit(1)      =", euler_limit(1))
print("euler_limit(10)     =", euler_limit(10))
print("euler_limit(1000)   =", euler_limit(1000))
print("euler_limit(10**6)  =", euler_limit(10**6))
print("真实 e              =", math.exp(1))

# 2. euler_infinite_sum：前 n 项（k 从 0 到 n-1）之和
def euler_infinite_sum(n):
    total = 0.0
    fact = 1              # 0! = 1
    for k in range(n):
        if k > 0:
            fact *= k      # k! = (k-1)! * k
        total += 1 / fact
    return total

print("euler_infinite_sum(0) =", euler_infinite_sum(0))   # 0
print("euler_infinite_sum(1) =", euler_infinite_sum(1))   # 1
print("euler_infinite_sum(4) =", euler_infinite_sum(4))   # ≈2.667
print("euler_infinite_sum(20)=", euler_infinite_sum(20))

# 3. euler_approx：累加直到与真实 e 的差 < epsilon
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) >= epsilon:
        if k > 0:
            fact *= k
        total += 1 / fact
        k += 1
    return total

for eps in [1e-1, 1e-3, 1e-6, 1e-10]:
    print(f"epsilon={eps:g}  ->  euler_approx = {euler_approx(eps)}")

# 4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"k={k:2d}  euler_infinite_sum = {euler_infinite_sum(k)}")

print_euler_sum_table(10)
```

四、测试整数属性

```python
# 1. is_power_of_2
def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

for x in [1, 2, 3, 4, 6, 8, 16, 18, 32, 100]:
    print(f"is_power_of_2({x}) = {is_power_of_2(x)}")

# 2. is_power：推广到任意底数 b
def is_power(b, n):
    if b < 2 or n < 1:
        return False
    if n == 1:          # b^0 = 1
        return True
    while n % b == 0:
        n //= b
    return n == 1

tests = [(2, 8), (2, 9), (3, 27), (3, 28), (5, 125), (5, 1), (10, 1000), (10, 1001)]
for b, n in tests:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```
