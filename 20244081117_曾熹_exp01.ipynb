实验 2：函数、条件、循环与递归

```python
#1
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须0‑100")
    if 90<=score<=100:
        return "优秀"
    elif 80<=score<90:
        return "良好"
    elif 60<=score<80:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(82))
print(score_level(70))
print(score_level(55))

#2 for 1‑100 sum
s_for = 0
for i in range(1,101):
    s_for +=i
print("for求和",s_for)

s_while=0
i=1
while i<=100:
    s_while +=i
    i +=1
print("while求和",s_while)

#3 阶乘 迭代、递归
def fact_iter(n):
    if n<0:
        raise ValueError
    res=1
    for x in range(2,n+1):
        res *=x
    return res

def fact_rec(n):
    if n<0:
        raise ValueError
    if n==0:
        return 1
    return n * fact_rec(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代={fact_iter(n)},递归={fact_rec(n)}")

#4 summary
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    mi = min(values)
    ma = max(values)
    avg = sum(values)/len(values)
    return (mi, ma, avg)

res = summary([10,20,30])
mi_,ma_,avg_ = res
print("元组解包：",mi_,ma_,avg_)

#5 sum_all 可变参数
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1,2,3,4))
```

实验 3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

#1 zip构建字典
stu_dict = dict(zip(names,scores))
for k,v in stu_dict.items():
    print(k,v)

#2 enumerate，序号从1
for idx,(n,s) in enumerate(zip(names,scores), start=1):
    print(idx,n,s)

#3列表推导式
pass_names = [n for n,s in zip(names,scores) if s>=60]
sq_scores = [s**2 for s in scores]
print("及格姓名",pass_names)
print("成绩平方",sq_scores)

#4统计
count = len(scores)
avg = sum(scores)/count
max_s = max(scores)
min_s = min(scores)
pass_cnt = sum(1 for s in scores if s>=60)
pass_rate = pass_cnt / count
print(f"人数{count},平均分{avg:.2f},最高{max_s},最低{min_s},及格率{pass_rate:.1%}")

#5 别名 vs copy
demo = [10,20,30]
alias = demo
copied = demo.copy()
alias[0]=999
print("原demo",demo)
print("alias",alias)
print("copied副本",copied)

#6元组不可修改，zip不等长
t = ("Tom",88)
try:
    t[1]=99
except TypeError as e:
    print("元组不能修改元素",e)

a1=[1,2]
b1=[10,20,30]
print("不等长zip",list(zip(a1,b1)))
```

实验 4 文件读写异常处理

```python
import os
import pickle

#1
print("工作目录",os.getcwd())
data_dir = os.path.join(os.getcwd(),"exp01_data")
os.makedirs(data_dir, exist_ok=True)

content_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

file_path = os.path.join(data_dir,"students.txt")
#2 write
with open(file_path,"w",encoding="utf‑8") as f:
    f.write(content_text)

#3 parse_student
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError("字段数量错误")
    name,s_str = parts
    name = name.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    score = int(s_str.strip())
    if not (0<=score<=100):
        raise ValueError(f"分数{score}不在0‑100")
    return {"name":name,"score":score}

#4逐行处理
valid_records = []
error_log = []
with open(file_path,"r",encoding="utf‑8") as f:
    lines = f.readlines()
for lineno,line in enumerate(lines, start=1):
    if lineno ==1:
        continue #跳过表头
    try:
        rec = parse_student(line)
        valid_records.append(rec)
    except Exception as e:
        error_log.append((lineno, str(e)))

print("有效记录：",valid_records)
print("错误记录行号原因：",error_log)

#5 report.txt w/a模式
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_sc = sum(r["score"] for r in valid_records)/valid_num
    p_count = sum(1 for r in valid_records if r["score"]>=60)
    p_rate = p_count / valid_num
else:
    avg_sc=0
    p_rate=0

report_path = os.path.join(data_dir,"report.txt")
# w模式覆盖
with open(report_path,"w",encoding="utf‑8") as f:
    f.write(f"有效记录数:{valid_num}\n")
    f.write(f"无效记录数:{invalid_num}\n")
    f.write(f"平均分:{avg_sc:.2f}\n")
    f.write(f"及格率:{p_rate:.2%}\n")

# a追加模式演示
with open(report_path,"a",encoding="utf‑8") as f:
    f.write("===追加的测试行===\n")

#6 pickle dump load
pkl_path = os.path.join(data_dir,"students.pkl")
with open(pkl_path,"wb") as fw:
    pickle.dump(valid_records, fw)

with open(pkl_path,"rb") as fr:
    loaded = pickle.load(fr)
print("pickle读取数据", loaded)
print("是否相等：", loaded == valid_records)

#8捕获文件不存在
try:
    with open("not_exist_1234.txt","r",encoding="utf‑8") as f:
        pass
except FileNotFoundError as e:
    print("捕获文件不存在异常：",e)
```

实验 5：类、继承、深浅拷贝

```python
import copy

#1 Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name={self.name},score={self.score},passed={self.is_passed()})"

#2从有效记录创建对象
for r in valid_records:
    s_obj = Student(r["name"], r["score"])
    print(s_obj)

#3 GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic},passed={self.is_passed()})"

g = GraduateStudent("LiHua", 82, "AI")
print(g)

#4 直接赋值 copy deepcopy
origin = {"name":"Alice", "scores":[85,90]}

#直接赋值
o1 = origin
o1["scores"][0]=999
print("直接赋值修改内层列表，origin变化",origin)

#恢复原始
origin = {"name":"Alice", "scores":[85,90]}
cp1 = copy.copy(origin)
cp1["scores"][0]= 777
print("copy.copy浅拷贝，内层列表会互相影响", origin, cp1)

origin = {"name":"Alice", "scores":[85,90]}
cp2 = copy.deepcopy(origin)
cp2["scores"][0] = 666
print("deepcopy深拷贝，互不影响", origin, cp2)
```
