正则表达式的基础匹配

```python
import re

# ========== 任务1：匹配数字、空白、单词边界 ==========
text1 = "cat on mat; later: cats and dogs"
print("文本:", text1)
print("\\d+ 匹配结果:", re.findall(r"\d+", text1))      
print("\\s+ 匹配结果:", re.findall(r"\s+", text1))     
print("\\bcat\\b 匹配结果:", re.findall(r"\bcat\b", text1)) 

# ========== 任务2：cat|dog 与 交替顺序差异 ==========
text2 = "cat dog bird CAT"
print("\n文本:", text2)
print("cat|dog 匹配:", re.findall(r"cat|dog", text2)) 

s = "aaaa"
print('re.findall(r"a|aa|aaa", "aaaa") =>', re.findall(r"a|aa|aaa", s))

print('re.findall(r"aaa|aa|a", "aaaa") =>', re.findall(r"aaa|aa|a", s))


# ========== 任务3：量词演示 ==========
text3 = "ct cat caat caaat"
print("\n文本:", text3)
print("ca*t  :", re.findall(r"ca*t", text3))  
print("ca+t  :", re.findall(r"ca+t", text3))  
print("ca?t  :", re.findall(r"ca?t", text3))  
print("ca{2}t:", re.findall(r"ca{2}t", text3))
print("ca{1,2}t:", re.findall(r"ca{1,2}t", text3)) 
```

分组、回溯引用与替换

```python
import re

# ========== 任务1：用捕获组提取邮箱的用户名和域名 ==========
text = "联系：alice@example.com；备选：bob@test.org "
pattern = r"([\w.]+)@([\w.]+)"
for user, domain in re.findall(pattern, text):
    print("用户名:", user, " 域名:", domain)

# ========== 任务2：后向引用找连续重复单词 ==========
text2 = "cat cat, dog cat, bird bird"

print("\n连续重复单词:", re.findall(r"\b(\w+)\s+\1\b", text2))

# ========== 任务3：日期格式转换 年-月-日 → 月-日-年 ==========
text3 = "2024-05-01\n2024/05/02\n2024 .05 .03"
print("\n转换前:\n" + text3)


result = re.sub(r"(\d{4})[-/.\s]*(\d{1,2})[-/.\s]*(\d{1,2})",
                r"\2-\3-\1", text3)
print("\n转换后:\n" + result)

# ========== 任务4：电话号码统一格式 ==========
text4 = "联系我：123-456-7890 或987 .654 .3210"
print("\n原始:", text4)

phone = re.sub(r"(\d{3})[-.\s]*(\d{3})[-.\s]*(\d{4})",
               r"(\1) \2-\3", text4)
print("统一后:", phone)
```

使用BeautifulSoup 解析HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06 ">讲义六</a>
<a href="https://example.org/lecture07 ">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

# ========== 任务1：页面标题、h1 标签名与文本 ==========
print("页面标题文本:", soup.title.get_text())
h1 = soup.find("h1")
print("h1 标签名:", h1.name)
print("h1 文本:", h1.get_text())

# ========== 任务2：find_all 找课程段落和链接 ==========
courses = soup.find_all("p", class_="course")
print("\n所有课程段落:")
for p in courses:
    print("  -", p.get_text())

links = soup.find_all("a")
print("\n所有链接:")
for a in links:
    print("  文字:", a.get_text(), " href:", a.get("href"))

# ========== 任务3：get_text() 可见文本比较 ==========
print("\n【整个文档可见文本】")
print(soup.get_text())

print("\n【单个段落可见文本】")
print(courses[0].get_text())

print("\n【单个链接可见文本】")
print(links[0].get_text())
```
