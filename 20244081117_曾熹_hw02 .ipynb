字符串的乐趣

```python
def is_palindrome(s):
   
    # 将字符串转换为小写并移除所有空格
    s = s.lower().replace(" ", "")
    # 比较字符串与其反转版本
    return s == s[::-1]


def is_abecedarian(s):
    # 移除空格
    s = s.replace(" ", "")
    
    # 遍历字符串，比较相邻字符
    for i in range(len(s) - 1):
        # 如果当前字符比下一个字符在字母表中靠后，则不是字母顺序
        if s[i] > s[i+1]:
            return False
            
    return True


def remove_vowels(s):
  
    vowels = "aeiou"
    result = ""
    
    # 遍历字符串中的每个字符
    for char in s:
        # 将字符转为小写进行判断，如果不是元音则保留
        if char.lower() not in vowels:
            result += char
            
    return result


# --- 测试代码 ---
if __name__ == "__main__":
    print("--- 测试 is_palindrome ---")
    print(f"is_palindrome('level'): {is_palindrome('level')}")                          
    print(f"is_palindrome('rats live on no evil star'): {is_palindrome('rats live on no evil star')}") 
    print(f"is_palindrome('T A C O cat'): {is_palindrome('T A C O cat')}")              
    print(f"is_palindrome('hello'): {is_palindrome('hello')}")                          

    print("\n--- 测试 is_abecedarian ---")
    print(f"is_abecedarian('adder'): {is_abecedarian('adder')}")
    print(f"is_abecedarian('cat'): {is_abecedarian('cat')}")                            
    print(f"is_abecedarian('abcd efgh xyz'): {is_abecedarian('abcd efgh xyz')}")        

    print("\n--- 测试 remove_vowels ---")
    print(f"remove_vowels('cat'): {remove_vowels('cat')}")
    print(f"remove_vowels('aeiou'): {remove_vowels('aeiou')}")                       
    print(f"remove_vowels('Hello World'): {remove_vowels('Hello World')}")             
```

列表的乐趣

```python
def list_reverse(lst):
    # 检查输入类型是否为列表
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表 (list)")
    
    # 使用切片 [::-1] 返回反转后的新列表
    return lst[::-1]


def is_sorted(seq):
    # 遍历序列，比较相邻元素
    # 对于长度为 0 或 1 的序列，range 为空，直接返回 True（符合升序定义）
    for i in range(len(seq) - 1):
        # 如果前一个元素大于后一个元素，说明不是升序
        if seq[i] > seq[i+1]:
            return False
            
    # 如果遍历完没有发现逆序对，则是升序
    return True


# --- 测试代码 ---
if __name__ == "__main__":
    print("--- 测试 list_reverse ---")
    print(f"list_reverse([1, 2, 3]): {list_reverse([1, 2, 3])}")      
    print(f"list_reverse(['a', 'b', 'c']): {list_reverse(['a', 'b', 'c'])}") 
    print(f"list_reverse([]): {list_reverse([])}")                 
    
    
    try:
        list_reverse("not a list")
    except TypeError as e:
        print(f"错误捕获成功: {e}")                                   

    print("\n--- 测试 is_sorted ---")
    print(f"is_sorted([1, 2, 2, 3]): {is_sorted([1, 2, 2, 3])}")      
    print(f"is_sorted([1, 3, 2, 4]): {is_sorted([1, 3, 2, 4])}")      
    print(f"is_sorted(['a', 'b', 'c']): {is_sorted(['a', 'b', 'c'])}") 
    print(f"is_sorted([5]): {is_sorted([5])}")                        
    print(f"is_sorted([]): {is_sorted([])}")                         
```

作为向量的元组

```python
def _is_number(x):
    return isinstance(x, (int, float)) and not isinstance(x, bool)


def vec_scalar_mult(t, s):
    
    # 类型检查：t 必须是元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组 (tuple)，表示向量")
    # 类型检查：标量 s 必须是数字
    if not _is_number(s):
        raise TypeError("第二个参数必须是整数或浮点数 (标量)")
    # 类型检查：t 中每个元素必须是数字
    for x in t:
        if not _is_number(x):
            raise TypeError("向量 t 中的元素必须都是整数或浮点数")
    
    # 返回新的元组
    return tuple(x * s for x in t)


def vec_inner_product(x, y):
 
    # 类型检查：两个参数都必须是元组
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个参数都必须是元组 (tuple)")
    
    # 长度检查：两个向量长度必须一致
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须一致")
    
    # 元素类型检查
    for v in x:
        if not _is_number(v):
            raise TypeError("向量 x 中的元素必须都是整数或浮点数")
    for v in y:
        if not _is_number(v):
            raise TypeError("向量 y 中的元素必须都是整数或浮点数")
    
    # 计算内积（强制返回浮点数）
    total = 0.0
    for xi, yi in zip(x, y):
        total += xi * yi
    return float(total)


def check_valid_mx(mx):

    # 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 空矩阵视为无效（矩阵至少应有 1 行 1 列；若不希望如此，可修改）
    if len(mx) == 0:
        return False
    # 每行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 每一行长度必须相等
    n = len(mx[0])
    for row in mx:
        if len(row) != n:
            return False
    # 每个元素必须是整数或浮点数
    for row in mx:
        for v in row:
            if not _is_number(v):
                return False
    return True


def mx_vec_mult(mx, v):
  
    # 检查矩阵有效性
    if not check_valid_mx(mx):
        raise TypeError("第一个参数必须是有效的矩阵（元组的元组，元素为数字）")
    # 检查向量有效性
    if not isinstance(v, tuple):
        raise TypeError("第二个参数必须是元组 (向量)")
    for x in v:
        if not _is_number(x):
            raise TypeError("向量 v 中的元素必须都是整数或浮点数")
    
    n = len(mx[0])
    if len(v) != n:
        raise ValueError(f"向量长度 {len(v)} 与矩阵列数 {n} 不匹配")
    
  
    result = []
    for row in mx:
        total = 0.0
        for a, b in zip(row, v):
            total += a * b
        result.append(total)
    return tuple(result)


# --- 测试代码 ---
if __name__ == "__main__":
    print("--- 测试 vec_scalar_mult ---")
    print(vec_scalar_mult((1, 2, 3), 2))        
    print(vec_scalar_mult((1.5, 2.5), 2))       
    try:
        vec_scalar_mult([1, 2, 3], 2)
    except TypeError as e:
        print("捕获 TypeError:", e)
    try:
        vec_scalar_mult((1, 2, 3), "a")
    except TypeError as e:
        print("捕获 TypeError:", e)

    print("\n--- 测试 vec_inner_product ---")
    print(vec_inner_product((1, 2, 3), (4, 5, 6)))  
    try:
        vec_inner_product((1, 2, 3), (4, 5))
    except ValueError as e:
        print("捕获 ValueError:", e)
    try:
        vec_inner_product([1, 2], (3, 4))
    except TypeError as e:
        print("捕获 TypeError:", e)

    print("\n--- 测试 check_valid_mx ---")
    print(check_valid_mx(((1, 2), (3, 4))))      
    print(check_valid_mx(((1, 2), (3,))))         
    print(check_valid_mx(((1, 2), [3, 4])))      
    print(check_valid_mx(((1, "a"), (3, 4))))     
    print(check_valid_mx([]))                     

    print("\n--- 测试 mx_vec_mult ---")
    mx = ((1, 2), (3, 4))
    v = (1, 1)
    print(mx_vec_mult(mx, v))                    
    try:
        mx_vec_mult(mx, (1, 2, 3))
    except ValueError as e:
        print("捕获 ValueError:", e)
    try:
        mx_vec_mult(((1, 2), (3,)), (1, 2))
    except TypeError as e:
        print("捕获 TypeError:", e)
```

更多向量的乐趣

```python
class Vector:
    def __init__(self, dimension, values=None):
     
        # 1. 检查维度类型是否为整数（排除 bool，因为 bool 是 int 的子类）
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("维度 (dimension) 必须是整数")
        
        # 2. 检查维度是否为非负
        if dimension < 0:
            raise ValueError("维度 (dimension) 不能为负数")
            
        self.dimension = dimension
        
        # 3. 处理未提供元素的情况（默认行为）
        if values is None:
            # 题目建议“选择合理的默认行为”，这里默认所有元素初始化为 0.0
            self.values = tuple(0.0 for _ in range(dimension))
        else:
            # 4. 检查元素类型是否为列表或元组
            if not isinstance(values, (list, tuple)):
                raise TypeError("元素 (values) 必须是列表 (list) 或元组 (tuple)")
                
            # 5. 检查维度与元素个数是否一致
            if len(values) != dimension:
                raise ValueError(f"元素的个数 ({len(values)}) 与指定的维度 ({dimension}) 不一致")
                
            # 6. 检查元素是否都为数字，并转换为元组存储（保证不可变性）
            for v in values:
                if not isinstance(v, (int, float)) or isinstance(v, bool):
                    raise TypeError("向量的元素必须都是整数或浮点数")
            self.values = tuple(values)

    def _check_comparable(self, other):
        #辅助方法：检查两个向量维度是否一致且对方是否为 Vector 对象 
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须是 Vector 的实例")
        if self.dimension != other.dimension:
            raise ValueError("不能比较两个维度不同的向量")

    # --- 比较运算符 ---

    def __eq__(self, other):
        #判断相等 (==) 
        if not isinstance(other, Vector):
            return False
        if self.dimension != other.dimension:
            return False  # Python 习惯，不同类型或维度直接返回 False 而不是抛出异常
        return self.values == other.values

    def __lt__(self, other):
        #判断小于 (<) 
        self._check_comparable(other)
        # 直接利用 Python 元组的字典序比较
        return self.values < other.values

    def __le__(self, other):
        #判断小于或等于 (<=) 
        self._check_comparable(other)
        return self.values <= other.values

    def __gt__(self, other):
       #判断大于 (>) 
        self._check_comparable(other)
        return self.values > other.values

    def __ge__(self, other):
        #判断大于或等于 (>=) 
        self._check_comparable(other)
        return self.values >= other.values

    # --- 数学运算 ---

    def dot(self, other):
        #计算两个向量的内积
        self._check_comparable(other)
        return sum(x * y for x, y in zip(self.values, other.values))

    def __mul__(self, other):
        #乘法
        # 1. 标量右乘：v * 2
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dimension, [x * other for x in self.values])
            
        # 2. 向量乘法（对应位置相乘）：v * w
        elif isinstance(other, Vector):
            self._check_comparable(other)
            return Vector(self.dimension, [x * y for x, y in zip(self.values, other.values)])
            
        else:
            raise TypeError("乘法操作数必须是数字 (标量) 或 Vector 对象")

    def __rmul__(self, other):
       #实现标量左乘：2 * v (当左侧不是 Vector 类型时，Python 会调用此方法)
        return self.__mul__(other)

    def __repr__(self):
        #定义打印输出格式，方便调试
        return f"Vector({self.dimension}, {self.values})"


# --- 测试代码 ---
if __name__ == "__main__":
    print("--- 测试初始化 ---")
    v1 = Vector(3, (2, 4, 6))
    v2 = Vector(3, [1, 2, 3])
    v_default = Vector(2)
    print(f"v1: {v1}, v2: {v2}, v_default: {v_default}")

    print("\n--- 测试错误处理 ---")
    try: Vector(-1, [])
    except ValueError as e: print("捕获 ValueError:", e)
    try: Vector(3, [1, 2])
    except ValueError as e: print("捕获 ValueError:", e)
    try: Vector("3", [1, 2, 3])
    except TypeError as e: print("捕获 TypeError:", e)

    print("\n--- 测试比较运算符 ---")
    v3 = Vector(2, (2, 4))
    v4 = Vector(2, (2, 5))
    print(f"v3 < v4: {v3 < v4}")      # True
    print(f"v3 >= v4: {v3 >= v4}")    # False
    print(f"v3 == Vector(2, (2,4)): {v3 == Vector(2, (2,4))}") # True
    
    try: v3 < Vector(3, (1, 2, 3))
    except ValueError as e: print("捕获 ValueError:", e)

    print("\n--- 测试内积 dot ---")
    print(f"v1.dot(v2): {v1.dot(v2)}")  # 2*1 + 4*2 + 6*3 = 28

    print("\n--- 测试乘法运算符 ---")
    print(f"v1 * 2: {v1 * 2}")          
    print(f"2 * v1: {2 * v1}")          
    print(f"v1 * v2: {v1 * v2}")        
```

关于问题2的回答：类属性还是实例属性？
维度和元素应作为实例属性。
理由： 每个具体的 Vector 对象（例如一个是二维的，一个是三维的）都有自己独立的维度和元素。如果作为类属性，所有的向量实例将共享同一个维度值和同一份元素数据，这显然违背了设计初衷。实例属性通过 self.dimension 和 self.values 绑定到具体的对象上，保证了每个向量独立拥有自己的状态。
